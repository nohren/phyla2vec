```python
import torch

def embed(x, dim):
    # Dummy embedding function
    N = dim  # Example size for the new dimension
    # expand x to have an additional dimension
    x_expanded = x.unsqueeze(-1).expand(-1, -1, N) # (B*R, L, N)
    
    noise = torch.randn(*x_expanded.shape)
    result = x_expanded + noise
    return result

# create tensor based on dim
a = torch.zeros(2,2,3, dtype=torch.float32)  # shape (B=2, R=2, L=3)
B,R,L = a.shape
print(f"Tensor starting shape with torch.tensor")
print(a.shape)
print(a)
b = a.view(B*R, L) #(B*R, L)
# print(f"Collapsed a dim with view")
# print(b.shape)
# print(b)

# # embed
# c = b.unsqueeze(-1)  # adds a new dimension of size 1 to a tensor at the specified position
# print(f"Unsqueezed b for embedding")
# print(c.shape)
# print(c)
# d = c.expand(-1,-1,5) # unsqueeze and expand go together, expand repeats the data along the new dimension, passing -1 for a dimension means to keep the original size in that dimension
# print(f"Expanded c for embedding")
# print(d.shape)
# print(d)
# noise = torch.randn(B*R,L, 3)
# assert noise.shape == d.shape, f"Noise shape {noise.shape} does not match expanded tensor shape {d.shape}"
# e = d + noise
# print(f"Added noise to expanded tensor")
# print(e.shape)
embedded = embed(b, dim=5)
print(f"Embedded tensor shape")
print(embedded.shape)
print(*embedded.shape)
```
